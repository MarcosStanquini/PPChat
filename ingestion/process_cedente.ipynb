```python
import os
from dotenv import load_dotenv

load_dotenv()

file_path = os.getenv('FILE_PATH')
```

```python
import pandas as pd
from typing import List, Dict, Any
from langchain_core.documents import Document
```

## Document Struct - LangChain

### *Metadata*



#### Example:

```python
def process_csv(filepath: str) -> List[Document]:
    df = pd.read_csv(file_path)
    
    df = df.fillna({
    'dataEntradaRJ': 'Não informado',
    'UF': 'Não informado',
    'nmRegiao': 'Não informado',
    'nmCidade': 'Não informado',
    'nmBairro': 'Não informado'
    })
    documents = []

    for idx, row in df.iterrows():
        content = f"""
        Informações do cedente:

        O cedente pertence ao segmento {row['nmSegmento']}, localizado em {row['nmBairro']}, {row['nmCidade']} - {row['UF']} ({row['nmRegiao']}).

        A validade do cadastro vai de {row['dataValidadeInicial']} até {row['dataValidadeFinal']}.  
        O limite global informado é de {row['valorLimiteGlobal']}.

        Situação judicial: {row['flagRecuperacaoJudicial']}.  
        Data de entrada em recuperação judicial (se aplicável): {row['dataEntradaRJ']}."""

        doc = Document(
            page_content=content,
            metadata = {
            "source": file_path,
            "row_index": idx,
            "idCedente": row["idCedente"],
            "idGrupoEconomico": row["idGrupoEconomico"],
            "UF": row["UF"],
            "nmCidade": row["nmCidade"],
            "nmRegiao": row["nmRegiao"],
            "nmBairro": row["nmBairro"],
            "nmSegmento": row["nmSegmento"],
            "flagRecuperacaoJudicial": row["flagRecuperacaoJudicial"],
            "valorLimiteGlobal": row["valorLimiteGlobal"],
            "dataCadastro": row["dataCadastro"],
            "dataValidadeInicial": row["dataValidadeInicial"],
            "dataValidadeFinal": row["dataValidadeFinal"],
            "dataEntradaRJ": row["dataEntradaRJ"],
        }

        )
        documents.append(doc)
    return documents
```

```python
process_csv(file_path)
```
